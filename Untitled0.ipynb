```python
# ============================================================
# PROJECT 10 — SMART-METER LOAD PROFILING
# UCI ElectricityLoadDiagrams20112014
# Complete Download + Extraction + CSV Conversion
# ============================================================

# If required, run this first:
# !pip install pandas requests tqdm

import os
import zipfile
import requests
import pandas as pd
from tqdm.notebook import tqdm


# ============================================================
# 1. CREATE FOLDERS
# ============================================================

RAW_DIR = "data/raw"
EXTRACT_DIR = "data/raw/uci_electricity"
PROCESSED_DIR = "data/processed"

os.makedirs(RAW_DIR, exist_ok=True)
os.makedirs(EXTRACT_DIR, exist_ok=True)
os.makedirs(PROCESSED_DIR, exist_ok=True)

print("Folders created successfully.")


# ============================================================
# 2. DOWNLOAD DATASET FROM UCI
# ============================================================

URL = (
    "https://archive.ics.uci.edu/static/public/321/"
    "electricityloaddiagrams20112014.zip"
)

ZIP_PATH = os.path.join(
    RAW_DIR,
    "electricityloaddiagrams20112014.zip"
)

print("\nDownloading UCI ElectricityLoadDiagrams20112014 dataset...")
print("This file is large, so the download may take some time.\n")

response = requests.get(URL, stream=True, timeout=120)
response.raise_for_status()

total_size = int(response.headers.get("content-length", 0))

with open(ZIP_PATH, "wb") as file:

    with tqdm(
        total=total_size,
        unit="B",
        unit_scale=True,
        unit_divisor=1024,
        desc="Downloading"
    ) as progress:

        for chunk in response.iter_content(
            chunk_size=1024 * 1024
        ):

            if chunk:
                file.write(chunk)
                progress.update(len(chunk))

print("\nDownload completed!")
print("Saved at:", ZIP_PATH)


# ============================================================
# 3. EXTRACT ZIP FILE
# ============================================================

print("\nExtracting dataset...")

with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
    zip_ref.extractall(EXTRACT_DIR)

print("Extraction completed!")


# ============================================================
# 4. FIND THE TXT FILE
# ============================================================

txt_file = None

for root, dirs, files in os.walk(EXTRACT_DIR):

    for filename in files:

        if filename.lower().endswith(".txt"):

            txt_file = os.path.join(root, filename)
            break

    if txt_file:
        break


if txt_file is None:
    raise FileNotFoundError(
        "Could not find the electricity dataset TXT file."
    )

print("\nDataset file found:")
print(txt_file)


# ============================================================
# 5. READ ORIGINAL 15-MINUTE DATA
# ============================================================

print("\nReading dataset...")
print("This may require several GB of RAM while processing.")

df = pd.read_csv(
    txt_file,
    sep=";",
    decimal=","
)

print("\nDataset loaded successfully!")


# ============================================================
# 6. DISPLAY BASIC INFORMATION
# ============================================================

print("\n================ DATASET INFORMATION ================\n")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print(
    "Number of clients:",
    df.shape[1] - 1
)

print("\nFirst 5 rows:")
print(df.head())


# ============================================================
# 7. RENAME TIMESTAMP COLUMN
# ============================================================

df.rename(
    columns={df.columns[0]: "timestamp"},
    inplace=True
)

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)

print("\nTimestamp converted successfully.")


# ============================================================
# 8. CHECK TIMESTAMP RANGE
# ============================================================

print("\n================ TIME RANGE ================\n")

print(
    "Start:",
    df["timestamp"].min()
)

print(
    "End:",
    df["timestamp"].max()
)


# ============================================================
# 9. CHECK NUMBER OF CLIENTS
# ============================================================

client_columns = list(df.columns[1:])

print(
    "\nNumber of clients:",
    len(client_columns)
)

print(
    "\nFirst 10 client IDs:"
)

print(
    client_columns[:10]
)


# ============================================================
# 10. CHECK MISSING VALUES
# ============================================================

print("\n================ MISSING VALUES ================\n")

total_missing = df.isna().sum().sum()

print(
    "Total missing values:",
    total_missing
)

if total_missing == 0:
    print("No missing values found.")
else:
    print("Missing values detected.")


# ============================================================
# 11. SAVE ORIGINAL DATA AS CSV
# ============================================================

ORIGINAL_CSV = os.path.join(
    PROCESSED_DIR,
    "electricity_load_2011_2014_15min.csv"
)

print("\nSaving original 15-minute data as CSV...")

df.to_csv(
    ORIGINAL_CSV,
    index=False
)

print(
    "Saved:",
    ORIGINAL_CSV
)


# ============================================================
# 12. CREATE HOURLY DATA
# ============================================================

print("\nCreating hourly dataset...")

hourly_df = df.copy()

hourly_df = hourly_df.set_index(
    "timestamp"
)

# Average four 15-minute readings into one hourly reading
hourly_df = hourly_df.resample(
    "1h"
).mean()

hourly_df = hourly_df.reset_index()

print(
    "Hourly dataset created."
)

print(
    "Hourly shape:",
    hourly_df.shape
)


# ============================================================
# 13. SAVE HOURLY CSV
# ============================================================

HOURLY_CSV = os.path.join(
    PROCESSED_DIR,
    "electricity_load_hourly.csv"
)

hourly_df.to_csv(
    HOURLY_CSV,
    index=False
)

print(
    "\nHourly CSV saved:"
)

print(
    HOURLY_CSV
)


# ============================================================
# 14. CREATE SMALL SAMPLE FOR TESTING
# ============================================================

print("\nCreating small sample dataset...")

sample_df = hourly_df[
    (hourly_df["timestamp"] >= "2011-01-01") &
    (hourly_df["timestamp"] < "2011-02-01")
].copy()

SAMPLE_CSV = os.path.join(
    PROCESSED_DIR,
    "electricity_sample_january_2011.csv"
)

sample_df.to_csv(
    SAMPLE_CSV,
    index=False
)

print(
    "Sample saved:",
    SAMPLE_CSV
)


# ============================================================
# 15. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("              DOWNLOAD COMPLETED")
print("=" * 60)

print("\nOriginal dataset:")
print(
    "Rows:",
    df.shape[0],
    "| Columns:",
    df.shape[1]
)

print(
    "Clients:",
    len(client_columns)
)

print(
    "Time range:",
    df["timestamp"].min(),
    "to",
    df["timestamp"].max()
)

print("\nHourly dataset:")
print(
    "Rows:",
    hourly_df.shape[0],
    "| Columns:",
    hourly_df.shape[1]
)

print("\nFiles created:")

print(
    "1.",
    ORIGINAL_CSV
)

print(
    "2.",
    HOURLY_CSV
)

print(
    "3.",
    SAMPLE_CSV
)

print("\nProject 10 data preparation completed!")
```

```python
# ============================================================
# PROJECT 10 — DATA QUALITY + BALANCE CHECK
# ElectricityLoadDiagrams20112014
# ============================================================

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD THE HOURLY DATASET
# ============================================================

file_path = "data/processed/electricity_load_hourly.csv"

df = pd.read_csv(file_path)

df["timestamp"] = pd.to_datetime(df["timestamp"])

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Number of meters:", df.shape[1] - 1)

print("\nTime range:")
print(df["timestamp"].min(), "to", df["timestamp"].max())


# ============================================================
# 2. CHECK DUPLICATE TIMESTAMPS
# ============================================================

print("\n" + "=" * 70)
print("1. DUPLICATE TIMESTAMP CHECK")
print("=" * 70)

duplicate_timestamps = df["timestamp"].duplicated().sum()

print("Duplicate timestamps:", duplicate_timestamps)

if duplicate_timestamps == 0:
    print("✓ No duplicate timestamps found.")
else:
    print("⚠ Duplicate timestamps found.")


# ============================================================
# 3. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("2. MISSING VALUE CHECK")
print("=" * 70)

missing_counts = df.isnull().sum()

total_missing = missing_counts.sum()

print("Total missing values:", total_missing)

if total_missing == 0:
    print("✓ No missing values found.")
else:
    print("⚠ Missing values found.")

    missing_table = pd.DataFrame({
        "missing_count": missing_counts,
        "missing_percentage": (
            missing_counts / len(df) * 100
        )
    })

    missing_table = missing_table[
        missing_table["missing_count"] > 0
    ]

    print("\nMeters/columns with missing values:")
    print(missing_table)


# ============================================================
# 4. CHECK INFINITE VALUES
# ============================================================

print("\n" + "=" * 70)
print("3. INFINITE VALUE CHECK")
print("=" * 70)

numeric_data = df.drop(columns=["timestamp"])

infinite_count = np.isinf(
    numeric_data.to_numpy()
).sum()

print("Infinite values:", infinite_count)

if infinite_count == 0:
    print("✓ No infinite values found.")
else:
    print("⚠ Infinite values found.")


# ============================================================
# 5. CHECK NEGATIVE CONSUMPTION VALUES
# ============================================================

print("\n" + "=" * 70)
print("4. NEGATIVE VALUE CHECK")
print("=" * 70)

negative_counts = (numeric_data < 0).sum()

total_negative = negative_counts.sum()

print("Total negative values:", total_negative)

if total_negative == 0:
    print("✓ No negative consumption values found.")
else:
    print("⚠ Negative consumption values found.")

    print(
        negative_counts[
            negative_counts > 0
        ]
    )


# ============================================================
# 6. CHECK ZERO VALUES
# ============================================================

print("\n" + "=" * 70)
print("5. ZERO CONSUMPTION CHECK")
print("=" * 70)

zero_counts = (numeric_data == 0).sum()

zero_percentage = (
    zero_counts / len(numeric_data) * 100
)

zero_table = pd.DataFrame({
    "zero_count": zero_counts,
    "zero_percentage": zero_percentage
})

zero_table = zero_table.sort_values(
    "zero_percentage",
    ascending=False
)

print("\nMeters with highest zero percentage:")

print(
    zero_table.head(20)
)


# ============================================================
# 7. IDENTIFY LEADING ZERO PERIOD
# ============================================================

print("\n" + "=" * 70)
print("6. LEADING ZERO PERIOD CHECK")
print("=" * 70)

leading_zero_results = []

for meter in numeric_data.columns:

    values = numeric_data[meter].to_numpy()

    non_zero_indices = np.where(values != 0)[0]

    if len(non_zero_indices) == 0:

        first_active_index = None
        leading_zero_count = len(values)

    else:

        first_active_index = non_zero_indices[0]
        leading_zero_count = first_active_index

    leading_zero_results.append({
        "meter": meter,
        "leading_zero_hours": leading_zero_count
    })


leading_zero_df = pd.DataFrame(
    leading_zero_results
)

leading_zero_df["leading_zero_days"] = (
    leading_zero_df["leading_zero_hours"] / 24
)

leading_zero_df = leading_zero_df.sort_values(
    "leading_zero_hours",
    ascending=False
)

print(
    leading_zero_df.head(20)
)


# ============================================================
# 8. CHECK DATA COVERAGE FOR EACH METER
# ============================================================

print("\n" + "=" * 70)
print("7. METER DATA COVERAGE")
print("=" * 70)

coverage_results = []

total_observations = len(df)

for meter in numeric_data.columns:

    values = numeric_data[meter]

    valid_count = values.notna().sum()

    coverage_percentage = (
        valid_count / total_observations
    ) * 100

    coverage_results.append({
        "meter": meter,
        "valid_observations": valid_count,
        "total_observations": total_observations,
        "coverage_percentage": coverage_percentage
    })


coverage_df = pd.DataFrame(
    coverage_results
)

print(
    coverage_df.head()
)

print(
    "\nMinimum coverage:",
    coverage_df["coverage_percentage"].min(),
    "%"
)

print(
    "Maximum coverage:",
    coverage_df["coverage_percentage"].max(),
    "%"
)


# ============================================================
# 9. DEFINE DATA-QUALITY STATUS
# ============================================================

# IMPORTANT:
# Since the dataset has no class labels, we do NOT use SMOTE.
#
# Instead, we identify meters that have poor observation
# coverage and may need exclusion from clustering.

MIN_COVERAGE = 90.0

coverage_df["quality_status"] = np.where(
    coverage_df["coverage_percentage"] >= MIN_COVERAGE,
    "Good",
    "Poor"
)

print("\n" + "=" * 70)
print("8. DATA QUALITY STATUS")
print("=" * 70)

print(
    coverage_df["quality_status"].value_counts()
)


# ============================================================
# 10. IDENTIFY METERS TO EXCLUDE
# ============================================================

poor_meters = coverage_df[
    coverage_df["quality_status"] == "Poor"
]["meter"].tolist()

good_meters = coverage_df[
    coverage_df["quality_status"] == "Good"
]["meter"].tolist()

print("\nGood-quality meters:", len(good_meters))
print("Poor-quality meters:", len(poor_meters))

if len(poor_meters) > 0:

    print("\nMeters recommended for inspection/exclusion:")

    print(poor_meters)

else:

    print(
        "\n✓ All meters satisfy the minimum coverage rule."
    )


# ============================================================
# 11. CREATE CLEAN DATASET
# ============================================================

clean_df = df[
    ["timestamp"] + good_meters
].copy()

print("\n" + "=" * 70)
print("9. CLEAN DATASET")
print("=" * 70)

print(
    "Original meters:",
    len(numeric_data.columns)
)

print(
    "Remaining meters:",
    len(good_meters)
)

print(
    "Removed meters:",
    len(poor_meters)
)

print(
    "Clean dataset shape:",
    clean_df.shape
)


# ============================================================
# 12. CHECK WHETHER THE DATASET IS CLASS-BALANCED
# ============================================================

print("\n" + "=" * 70)
print("10. CLASS BALANCE CHECK")
print("=" * 70)

print("""
IMPORTANT:
This dataset is NOT a classification dataset.

There is:
    - No target variable
    - No class column
    - No labels such as Home / Shop / Office / Factory

Therefore:

    Accuracy / class balance / SMOTE
    cannot be applied to this raw dataset.

This is an UNSUPERVISED learning problem.

The appropriate next step is CUSTOMER LOAD-PROFILE
CLUSTERING, not SMOTE.
""")


# ============================================================
# 13. CHECK CUSTOMER LOAD SIZE DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("11. CUSTOMER LOAD DISTRIBUTION")
print("=" * 70)

meter_means = clean_df[
    good_meters
].mean()

meter_stats = pd.DataFrame({
    "meter": meter_means.index,
    "mean_consumption": meter_means.values
})

print(
    meter_stats.describe()
)


# ============================================================
# 14. IDENTIFY VERY SMALL / VERY LARGE METERS
# ============================================================

q1 = meter_stats[
    "mean_consumption"
].quantile(0.25)

q3 = meter_stats[
    "mean_consumption"
].quantile(0.75)

iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

size_outliers = meter_stats[
    (meter_stats["mean_consumption"] < lower_limit) |
    (meter_stats["mean_consumption"] > upper_limit)
]

print("\nPotential load-size outliers:")
print(size_outliers)


# ============================================================
# 15. SAVE QUALITY REPORT
# ============================================================

os.makedirs(
    "outputs",
    exist_ok=True
)

coverage_df.to_csv(
    "outputs/meter_coverage_report.csv",
    index=False
)

leading_zero_df.to_csv(
    "outputs/leading_zero_report.csv",
    index=False
)

zero_table.to_csv(
    "outputs/zero_consumption_report.csv"
)

meter_stats.to_csv(
    "outputs/meter_load_statistics.csv",
    index=False
)

clean_df.to_csv(
    "data/processed/electricity_load_hourly_clean.csv",
    index=False
)


# ============================================================
# 16. VISUALIZE CUSTOMER LOAD SIZE
# ============================================================

plt.figure(figsize=(10, 5))

plt.hist(
    meter_stats["mean_consumption"],
    bins=30
)

plt.xlabel("Average Consumption")
plt.ylabel("Number of Meters")
plt.title("Distribution of Average Customer Consumption")

plt.tight_layout()
plt.show()


# ============================================================
# 17. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("FINAL DATA QUALITY SUMMARY")
print("=" * 70)

print("Total meters:", len(numeric_data.columns))

print(
    "Total missing values:",
    total_missing
)

print(
    "Infinite values:",
    infinite_count
)

print(
    "Negative values:",
    total_negative
)

print(
    "Meters passing coverage rule:",
    len(good_meters)
)

print(
    "Meters requiring inspection:",
    len(poor_meters)
)

print("\nSMOTE required:", "NO")

print("""
Reason:
This is an UNSUPERVISED customer load-profile problem.
There are no target classes to balance.

Next step:
Feature engineering → normalization → clustering.
""")

print("=" * 70)
print("DATA QUALITY CHECK COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# PROJECT 10 - SMART METER LOAD PROFILING
# EDA + 10 VISUALIZATIONS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

FILE_PATH = "data/processed/electricity_load_hourly.csv"

df = pd.read_csv(FILE_PATH)

df["timestamp"] = pd.to_datetime(df["timestamp"])

# Customer columns
meters = [col for col in df.columns if col != "timestamp"]

print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Number of meters:", len(meters))
print("Start:", df["timestamp"].min())
print("End:", df["timestamp"].max())

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------------------
# 2. CREATE OUTPUT FOLDER
# ------------------------------------------------------------

os.makedirs("outputs/eda", exist_ok=True)


# ============================================================
# GRAPH 1 - TOTAL SYSTEM LOAD OVER TIME
# ============================================================

df["total_load"] = df[meters].sum(axis=1)

plt.figure(figsize=(14, 5))

plt.plot(
    df["timestamp"],
    df["total_load"],
    linewidth=0.8
)

plt.title("Total System Electricity Load Over Time")
plt.xlabel("Date")
plt.ylabel("Total Load (kW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/01_total_system_load.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 2 - AVERAGE HOURLY LOAD PROFILE
# ============================================================

df["hour"] = df["timestamp"].dt.hour

hourly_profile = (
    df.groupby("hour")["total_load"]
    .mean()
)

plt.figure(figsize=(10, 5))

plt.plot(
    hourly_profile.index,
    hourly_profile.values,
    marker="o"
)

plt.title("Average Daily Load Profile")
plt.xlabel("Hour of Day")
plt.ylabel("Average Load (kW)")
plt.xticks(range(24))
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/02_average_daily_profile.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 3 - WEEKDAY VS WEEKEND LOAD
# ============================================================

df["day_of_week"] = df["timestamp"].dt.dayofweek

df["day_type"] = np.where(
    df["day_of_week"] < 5,
    "Weekday",
    "Weekend"
)

weekday_profile = (
    df.groupby(
        ["day_type", "hour"]
    )["total_load"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 5))

for day_type in ["Weekday", "Weekend"]:

    temp = weekday_profile[
        weekday_profile["day_type"] == day_type
    ]

    plt.plot(
        temp["hour"],
        temp["total_load"],
        marker="o",
        label=day_type
    )

plt.title("Weekday vs Weekend Load Profile")
plt.xlabel("Hour of Day")
plt.ylabel("Average Load (kW)")
plt.xticks(range(24))
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/03_weekday_vs_weekend.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 4 - MONTHLY AVERAGE LOAD
# ============================================================

df["month"] = df["timestamp"].dt.month

monthly_load = (
    df.groupby("month")["total_load"]
    .mean()
)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_load.index,
    monthly_load.values,
    marker="o"
)

plt.title("Average Monthly Electricity Load")
plt.xlabel("Month")
plt.ylabel("Average Load (kW)")
plt.xticks(range(1, 13))
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/04_monthly_load.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 5 - DISTRIBUTION OF CUSTOMER AVERAGE LOAD
# ============================================================

customer_mean = df[meters].mean()

plt.figure(figsize=(10, 5))

plt.hist(
    customer_mean,
    bins=30
)

plt.title("Distribution of Average Customer Load")
plt.xlabel("Average Consumption (kW)")
plt.ylabel("Number of Customers")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/05_customer_average_distribution.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 6 - TOP 20 CUSTOMERS BY AVERAGE CONSUMPTION
# ============================================================

top_customers = (
    customer_mean
    .sort_values(ascending=False)
    .head(20)
)

plt.figure(figsize=(12, 6))

plt.bar(
    top_customers.index,
    top_customers.values
)

plt.title("Top 20 Customers by Average Electricity Consumption")
plt.xlabel("Customer Meter")
plt.ylabel("Average Load (kW)")
plt.xticks(rotation=90)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/06_top_20_customers.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 7 - TOP 20 CUSTOMERS BY PEAK LOAD
# ============================================================

customer_peak = df[meters].max()

top_peak_customers = (
    customer_peak
    .sort_values(ascending=False)
    .head(20)
)

plt.figure(figsize=(12, 6))

plt.bar(
    top_peak_customers.index,
    top_peak_customers.values
)

plt.title("Top 20 Customers by Peak Load")
plt.xlabel("Customer Meter")
plt.ylabel("Peak Load (kW)")
plt.xticks(rotation=90)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/07_top_20_peak_customers.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 8 - LOAD FACTOR DISTRIBUTION
# ============================================================

customer_mean = df[meters].mean()
customer_peak = df[meters].max()

load_factor = (
    customer_mean / customer_peak
)

load_factor = load_factor.replace(
    [np.inf, -np.inf],
    np.nan
).dropna()

plt.figure(figsize=(10, 5))

plt.hist(
    load_factor,
    bins=30
)

plt.title("Distribution of Customer Load Factor")
plt.xlabel("Load Factor")
plt.ylabel("Number of Customers")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/08_load_factor_distribution.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 9 - CUSTOMER AVERAGE LOAD VS PEAK LOAD
# ============================================================

customer_statistics = pd.DataFrame({
    "average_load": customer_mean,
    "peak_load": customer_peak
})

plt.figure(figsize=(10, 6))

plt.scatter(
    customer_statistics["average_load"],
    customer_statistics["peak_load"],
    alpha=0.7
)

plt.title("Average Load vs Peak Load Across Customers")
plt.xlabel("Average Load (kW)")
plt.ylabel("Peak Load (kW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "outputs/eda/09_average_vs_peak.png",
    dpi=300
)

plt.show()


# ============================================================
# GRAPH 10 - HEATMAP OF AVERAGE HOURLY LOAD BY MONTH
# ============================================================

df["month_name"] = df["timestamp"].dt.month

monthly_hourly = (
    df.groupby(
        ["month_name", "hour"]
    )["total_load"]
    .mean()
    .unstack()
)

plt.figure(figsize=(14, 6))

plt.imshow(
    monthly_hourly,
    aspect="auto"
)

plt.colorbar(
    label="Average Load (kW)"
)

plt.title(
    "Average Electricity Load by Month and Hour"
)

plt.xlabel("Hour of Day")
plt.ylabel("Month")

plt.xticks(
    range(24),
    range(24)
)

plt.yticks(
    range(12),
    range(1, 13)
)

plt.tight_layout()

plt.savefig(
    "outputs/eda/10_month_hour_heatmap.png",
    dpi=300
)

plt.show()


# ============================================================
# 3. BASIC EDA STATISTICS
# ============================================================

print("\n")
print("=" * 70)
print("EDA SUMMARY")
print("=" * 70)

print("\nTotal system average load:")
print(
    round(
        df["total_load"].mean(),
        2
    ),
    "kW"
)

print("\nTotal system maximum load:")
print(
    round(
        df["total_load"].max(),
        2
    ),
    "kW"
)

print("\nCustomer average load:")
print(
    round(
        customer_mean.mean(),
        2
    ),
    "kW"
)

print("\nCustomer maximum average load:")
print(
    round(
        customer_mean.max(),
        2
    ),
    "kW"
)

print("\nAverage customer load factor:")
print(
    round(
        load_factor.mean(),
        3
    )
)


# ============================================================
# 4. EVENING PEAK ANALYSIS
# ============================================================

evening_data = df[
    df["hour"].between(17, 21)
]

evening_average = (
    evening_data["total_load"].mean()
)

system_peak = df["total_load"].max()

print("\n")
print("=" * 70)
print("EVENING PEAK ANALYSIS")
print("=" * 70)

print(
    "Evening peak average load (17:00-21:00):",
    round(evening_average, 2),
    "kW"
)

print(
    "Maximum system load:",
    round(system_peak, 2),
    "kW"
)


# ============================================================
# 5. SAVE CUSTOMER STATISTICS
# ============================================================

customer_statistics["load_factor"] = load_factor

customer_statistics.to_csv(
    "outputs/eda/customer_statistics.csv"
)

print("\nCustomer statistics saved.")


# ============================================================
# 6. SAVE HOURLY PROFILE
# ============================================================

hourly_profile.to_csv(
    "outputs/eda/hourly_system_profile.csv"
)

print("Hourly profile saved.")


# ============================================================
# FINAL MESSAGE
# ============================================================

print("\n")
print("=" * 70)
print("EDA COMPLETED SUCCESSFULLY")
print("=" * 70)

print("""
10 graphs generated:

1. Total system load over time
2. Average daily load profile
3. Weekday vs weekend load
4. Monthly average load
5. Customer average-load distribution
6. Top 20 customers by average consumption
7. Top 20 customers by peak load
8. Load-factor distribution
9. Average load vs peak load
10. Month-hour load heatmap

All graphs are saved in:

outputs/eda/
""")
```

```python
# ================================================================
# SMART-METER FEATURE ENGINEERING + FEATURE SELECTION
# Dataset: Electricity Load Diagrams
# ================================================================

import pandas as pd
import numpy as np
import os

from sklearn.feature_selection import VarianceThreshold
from sklearn.preprocessing import StandardScaler

# ================================================================
# 1. LOAD DATA
# ================================================================

file_path = "data/processed/electricity_load_hourly_clean.csv"

df = pd.read_csv(file_path)

df["timestamp"] = pd.to_datetime(df["timestamp"])

meter_cols = [c for c in df.columns if c.startswith("MT_")]

print("Dataset shape:", df.shape)
print("Number of meters:", len(meter_cols))


# ================================================================
# 2. CREATE TIME FEATURES
# ================================================================

df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["month"] = df["timestamp"].dt.month

df["is_weekend"] = df["day_of_week"] >= 5

# Evening = 5 PM to 9 PM
df["is_evening"] = df["hour"].between(17, 21)

# Night = 12 AM to 6 AM
df["is_night"] = df["hour"].between(0, 5)

# Morning = 6 AM to 11 AM
df["is_morning"] = df["hour"].between(6, 11)

# Afternoon = 12 PM to 4 PM
df["is_afternoon"] = df["hour"].between(12, 16)


# ================================================================
# 3. FEATURE ENGINEERING
# ================================================================

features = []

for meter in meter_cols:

    x = df[meter]

    feature = {}

    # ------------------------------------------------------------
    # BASIC STATISTICS
    # ------------------------------------------------------------

    feature["mean_load"] = x.mean()
    feature["median_load"] = x.median()
    feature["std_load"] = x.std()
    feature["min_load"] = x.min()
    feature["max_load"] = x.max()

    feature["range_load"] = x.max() - x.min()

    feature["variance_load"] = x.var()

    # ------------------------------------------------------------
    # PERCENTILES
    # ------------------------------------------------------------

    feature["p10_load"] = x.quantile(0.10)
    feature["p25_load"] = x.quantile(0.25)
    feature["p50_load"] = x.quantile(0.50)
    feature["p75_load"] = x.quantile(0.75)
    feature["p90_load"] = x.quantile(0.90)
    feature["p95_load"] = x.quantile(0.95)

    # ------------------------------------------------------------
    # LOAD FACTOR
    # ------------------------------------------------------------

    if x.max() > 0:
        feature["load_factor"] = x.mean() / x.max()
    else:
        feature["load_factor"] = 0

    # ------------------------------------------------------------
    # TIME-OF-DAY FEATURES
    # ------------------------------------------------------------

    feature["morning_mean"] = x[df["is_morning"]].mean()

    feature["afternoon_mean"] = x[df["is_afternoon"]].mean()

    feature["evening_mean"] = x[df["is_evening"]].mean()

    feature["night_mean"] = x[df["is_night"]].mean()

    # ------------------------------------------------------------
    # EVENING PEAK SHARE
    # ------------------------------------------------------------

    total_energy = x.sum()

    if total_energy > 0:
        feature["evening_energy_share"] = (
            x[df["is_evening"]].sum() / total_energy
        )
    else:
        feature["evening_energy_share"] = 0

    # ------------------------------------------------------------
    # NIGHT ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["night_energy_share"] = (
            x[df["is_night"]].sum() / total_energy
        )
    else:
        feature["night_energy_share"] = 0

    # ------------------------------------------------------------
    # MORNING ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["morning_energy_share"] = (
            x[df["is_morning"]].sum() / total_energy
        )
    else:
        feature["morning_energy_share"] = 0

    # ------------------------------------------------------------
    # AFTERNOON ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["afternoon_energy_share"] = (
            x[df["is_afternoon"]].sum() / total_energy
        )
    else:
        feature["afternoon_energy_share"] = 0

    # ------------------------------------------------------------
    # WEEKDAY / WEEKEND
    # ------------------------------------------------------------

    weekday_values = x[~df["is_weekend"]]
    weekend_values = x[df["is_weekend"]]

    feature["weekday_mean"] = weekday_values.mean()

    feature["weekend_mean"] = weekend_values.mean()

    if feature["weekday_mean"] > 0:
        feature["weekend_weekday_ratio"] = (
            feature["weekend_mean"] /
            feature["weekday_mean"]
        )
    else:
        feature["weekend_weekday_ratio"] = 0

    # ------------------------------------------------------------
    # PEAK TIMING
    # ------------------------------------------------------------

    hourly_mean = (
        pd.DataFrame({
            "load": x,
            "hour": df["hour"]
        })
        .groupby("hour")["load"]
        .mean()
    )

    peak_hour = hourly_mean.idxmax()

    feature["peak_hour"] = peak_hour

    # ------------------------------------------------------------
    # PEAK HOUR LOAD
    # ------------------------------------------------------------

    feature["peak_hour_load"] = hourly_mean.max()

    # ------------------------------------------------------------
    # PEAK / AVERAGE RATIO
    # ------------------------------------------------------------

    if feature["mean_load"] > 0:
        feature["peak_to_average_ratio"] = (
            feature["max_load"] /
            feature["mean_load"]
        )
    else:
        feature["peak_to_average_ratio"] = 0

    # ------------------------------------------------------------
    # COEFFICIENT OF VARIATION
    # ------------------------------------------------------------

    if feature["mean_load"] > 0:
        feature["coefficient_variation"] = (
            feature["std_load"] /
            feature["mean_load"]
        )
    else:
        feature["coefficient_variation"] = 0

    # ------------------------------------------------------------
    # ZERO CONSUMPTION
    # ------------------------------------------------------------

    feature["zero_consumption_pct"] = (
        (x == 0).mean() * 100
    )

    # ------------------------------------------------------------
    # HIGH LOAD FREQUENCY
    # ------------------------------------------------------------

    p90 = x.quantile(0.90)

    feature["high_load_frequency"] = (
        (x >= p90).mean()
    )

    # ------------------------------------------------------------
    # VOLATILITY
    # ------------------------------------------------------------

    feature["mean_absolute_change"] = (
        x.diff().abs().mean()
    )

    # ------------------------------------------------------------
    # CUSTOMER ID
    # ------------------------------------------------------------

    feature["customer_id"] = meter

    features.append(feature)


# ================================================================
# 4. CREATE CUSTOMER FEATURE DATASET
# ================================================================

features_df = pd.DataFrame(features)

# Put customer_id first
cols = ["customer_id"] + [
    c for c in features_df.columns
    if c != "customer_id"
]

features_df = features_df[cols]

print("\n" + "=" * 60)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 60)

print("Customers:", features_df.shape[0])
print("Features :", features_df.shape[1] - 1)

print("\nFeature names:")
for i, col in enumerate(features_df.columns[1:], 1):
    print(i, col)


# ================================================================
# 5. SAVE ALL FEATURES
# ================================================================

os.makedirs("outputs/feature_selection", exist_ok=True)

features_df.to_csv(
    "outputs/customer_features.csv",
    index=False
)

print("\nSaved:")
print("outputs/customer_features.csv")


# ================================================================
# 6. PREPARE DATA FOR FEATURE SELECTION
# ================================================================

X = features_df.drop(
    columns=["customer_id"]
)

# Replace infinity
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Fill missing values
X = X.fillna(X.median())


# ================================================================
# 7. REMOVE LOW-VARIANCE FEATURES
# ================================================================

print("\n" + "=" * 60)
print("FEATURE SELECTION - VARIANCE FILTER")
print("=" * 60)

# Standardize before variance calculation
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

# Very small variance threshold
variance_threshold = 0.01

selector = VarianceThreshold(
    threshold=variance_threshold
)

selector.fit(X_scaled_df)

variance_features = X.columns[
    selector.get_support()
].tolist()

removed_variance = [
    col for col in X.columns
    if col not in variance_features
]

X_variance = X[variance_features]

print("Original features:",
      X.shape[1])

print("Removed low variance:",
      len(removed_variance))

print("Remaining features:",
      X_variance.shape[1])


# ================================================================
# 8. CORRELATION FILTER
# ================================================================

print("\n" + "=" * 60)
print("FEATURE SELECTION - CORRELATION FILTER")
print("=" * 60)

corr_matrix = X_variance.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

correlation_threshold = 0.90

high_corr_features = [
    column
    for column in upper_triangle.columns
    if any(
        upper_triangle[column] >
        correlation_threshold
    )
]

X_selected = X_variance.drop(
    columns=high_corr_features
)

print("Before correlation filter:",
      X_variance.shape[1])

print("Highly correlated removed:",
      len(high_corr_features))

print("Final selected features:",
      X_selected.shape[1])


# ================================================================
# 9. DISPLAY FINAL FEATURES
# ================================================================

print("\n" + "=" * 60)
print("FINAL SELECTED FEATURES")
print("=" * 60)

for i, feature in enumerate(
    X_selected.columns,
    1
):
    print(f"{i}. {feature}")


# ================================================================
# 10. SAVE SELECTED FEATURES
# ================================================================

selected_features_df = pd.concat(
    [
        features_df[["customer_id"]],
        X_selected
    ],
    axis=1
)

selected_features_df.to_csv(
    "outputs/feature_selection/selected_features.csv",
    index=False
)


# ================================================================
# 11. CREATE FEATURE SELECTION REPORT
# ================================================================

report = pd.DataFrame({
    "feature": X.columns,

    "passed_variance_filter": [
        f in variance_features
        for f in X.columns
    ],

    "removed_by_correlation": [
        f in high_corr_features
        for f in X.columns
    ],

    "final_selected": [
        f in X_selected.columns
        for f in X.columns
    ]
})

report.to_csv(
    "outputs/feature_selection/feature_selection_report.csv",
    index=False
)


# ================================================================
# 12. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print(
    "Original features       :",
    X.shape[1]
)

print(
    "After variance filtering:",
    X_variance.shape[1]
)

print(
    "After correlation filter:",
    X_selected.shape[1]
)

print("\nFiles created:")

print(
    "1. outputs/customer_features.csv"
)

print(
    "2. outputs/feature_selection/selected_features.csv"
)

print(
    "3. outputs/feature_selection/feature_selection_report.csv"
)

print("\nFeature selection completed successfully!")
```

```python
# ================================================================
# SMART-METER FEATURE ENGINEERING + FEATURE SELECTION
# Dataset: Electricity Load Diagrams
# ================================================================

import pandas as pd
import numpy as np
import os

from sklearn.feature_selection import VarianceThreshold
from sklearn.preprocessing import StandardScaler

# ================================================================
# 1. LOAD DATA
# ================================================================

file_path = "data/processed/electricity_load_hourly_clean.csv"

df = pd.read_csv(file_path)

df["timestamp"] = pd.to_datetime(df["timestamp"])

meter_cols = [c for c in df.columns if c.startswith("MT_")]

print("Dataset shape:", df.shape)
print("Number of meters:", len(meter_cols))


# ================================================================
# 2. CREATE TIME FEATURES
# ================================================================

df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["month"] = df["timestamp"].dt.month

df["is_weekend"] = df["day_of_week"] >= 5

# Evening = 5 PM to 9 PM
df["is_evening"] = df["hour"].between(17, 21)

# Night = 12 AM to 6 AM
df["is_night"] = df["hour"].between(0, 5)

# Morning = 6 AM to 11 AM
df["is_morning"] = df["hour"].between(6, 11)

# Afternoon = 12 PM to 4 PM
df["is_afternoon"] = df["hour"].between(12, 16)


# ================================================================
# 3. FEATURE ENGINEERING
# ================================================================

features = []

for meter in meter_cols:

    x = df[meter]

    feature = {}

    # ------------------------------------------------------------
    # BASIC STATISTICS
    # ------------------------------------------------------------

    feature["mean_load"] = x.mean()
    feature["median_load"] = x.median()
    feature["std_load"] = x.std()
    feature["min_load"] = x.min()
    feature["max_load"] = x.max()

    feature["range_load"] = x.max() - x.min()

    feature["variance_load"] = x.var()

    # ------------------------------------------------------------
    # PERCENTILES
    # ------------------------------------------------------------

    feature["p10_load"] = x.quantile(0.10)
    feature["p25_load"] = x.quantile(0.25)
    feature["p50_load"] = x.quantile(0.50)
    feature["p75_load"] = x.quantile(0.75)
    feature["p90_load"] = x.quantile(0.90)
    feature["p95_load"] = x.quantile(0.95)

    # ------------------------------------------------------------
    # LOAD FACTOR
    # ------------------------------------------------------------

    if x.max() > 0:
        feature["load_factor"] = x.mean() / x.max()
    else:
        feature["load_factor"] = 0

    # ------------------------------------------------------------
    # TIME-OF-DAY FEATURES
    # ------------------------------------------------------------

    feature["morning_mean"] = x[df["is_morning"]].mean()

    feature["afternoon_mean"] = x[df["is_afternoon"]].mean()

    feature["evening_mean"] = x[df["is_evening"]].mean()

    feature["night_mean"] = x[df["is_night"]].mean()

    # ------------------------------------------------------------
    # EVENING PEAK SHARE
    # ------------------------------------------------------------

    total_energy = x.sum()

    if total_energy > 0:
        feature["evening_energy_share"] = (
            x[df["is_evening"]].sum() / total_energy
        )
    else:
        feature["evening_energy_share"] = 0

    # ------------------------------------------------------------
    # NIGHT ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["night_energy_share"] = (
            x[df["is_night"]].sum() / total_energy
        )
    else:
        feature["night_energy_share"] = 0

    # ------------------------------------------------------------
    # MORNING ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["morning_energy_share"] = (
            x[df["is_morning"]].sum() / total_energy
        )
    else:
        feature["morning_energy_share"] = 0

    # ------------------------------------------------------------
    # AFTERNOON ENERGY SHARE
    # ------------------------------------------------------------

    if total_energy > 0:
        feature["afternoon_energy_share"] = (
            x[df["is_afternoon"]].sum() / total_energy
        )
    else:
        feature["afternoon_energy_share"] = 0

    # ------------------------------------------------------------
    # WEEKDAY / WEEKEND
    # ------------------------------------------------------------

    weekday_values = x[~df["is_weekend"]]
    weekend_values = x[df["is_weekend"]]

    feature["weekday_mean"] = weekday_values.mean()

    feature["weekend_mean"] = weekend_values.mean()

    if feature["weekday_mean"] > 0:
        feature["weekend_weekday_ratio"] = (
            feature["weekend_mean"] /
            feature["weekday_mean"]
        )
    else:
        feature["weekend_weekday_ratio"] = 0

    # ------------------------------------------------------------
    # PEAK TIMING
    # ------------------------------------------------------------

    hourly_mean = (
        pd.DataFrame({
            "load": x,
            "hour": df["hour"]
        })
        .groupby("hour")["load"]
        .mean()
    )

    peak_hour = hourly_mean.idxmax()

    feature["peak_hour"] = peak_hour

    # ------------------------------------------------------------
    # PEAK HOUR LOAD
    # ------------------------------------------------------------

    feature["peak_hour_load"] = hourly_mean.max()

    # ------------------------------------------------------------
    # PEAK / AVERAGE RATIO
    # ------------------------------------------------------------

    if feature["mean_load"] > 0:
        feature["peak_to_average_ratio"] = (
            feature["max_load"] /
            feature["mean_load"]
        )
    else:
        feature["peak_to_average_ratio"] = 0

    # ------------------------------------------------------------
    # COEFFICIENT OF VARIATION
    # ------------------------------------------------------------

    if feature["mean_load"] > 0:
        feature["coefficient_variation"] = (
            feature["std_load"] /
            feature["mean_load"]
        )
    else:
        feature["coefficient_variation"] = 0

    # ------------------------------------------------------------
    # ZERO CONSUMPTION
    # ------------------------------------------------------------

    feature["zero_consumption_pct"] = (
        (x == 0).mean() * 100
    )

    # ------------------------------------------------------------
    # HIGH LOAD FREQUENCY
    # ------------------------------------------------------------

    p90 = x.quantile(0.90)

    feature["high_load_frequency"] = (
        (x >= p90).mean()
    )

    # ------------------------------------------------------------
    # VOLATILITY
    # ------------------------------------------------------------

    feature["mean_absolute_change"] = (
        x.diff().abs().mean()
    )

    # ------------------------------------------------------------
    # CUSTOMER ID
    # ------------------------------------------------------------

    feature["customer_id"] = meter

    features.append(feature)


# ================================================================
# 4. CREATE CUSTOMER FEATURE DATASET
# ================================================================

features_df = pd.DataFrame(features)

# Put customer_id first
cols = ["customer_id"] + [
    c for c in features_df.columns
    if c != "customer_id"
]

features_df = features_df[cols]

print("\n" + "=" * 60)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 60)

print("Customers:", features_df.shape[0])
print("Features :", features_df.shape[1] - 1)

print("\nFeature names:")
for i, col in enumerate(features_df.columns[1:], 1):
    print(i, col)


# ================================================================
# 5. SAVE ALL FEATURES
# ================================================================

os.makedirs("outputs/feature_selection", exist_ok=True)

features_df.to_csv(
    "outputs/customer_features.csv",
    index=False
)

print("\nSaved:")
print("outputs/customer_features.csv")


# ================================================================
# 6. PREPARE DATA FOR FEATURE SELECTION
# ================================================================

X = features_df.drop(
    columns=["customer_id"]
)

# Replace infinity
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Fill missing values
X = X.fillna(X.median())


# ================================================================
# 7. REMOVE LOW-VARIANCE FEATURES
# ================================================================

print("\n" + "=" * 60)
print("FEATURE SELECTION - VARIANCE FILTER")
print("=" * 60)

# Standardize before variance calculation
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

# Very small variance threshold
variance_threshold = 0.01

selector = VarianceThreshold(
    threshold=variance_threshold
)

selector.fit(X_scaled_df)

variance_features = X.columns[
    selector.get_support()
].tolist()

removed_variance = [
    col for col in X.columns
    if col not in variance_features
]

X_variance = X[variance_features]

print("Original features:",
      X.shape[1])

print("Removed low variance:",
      len(removed_variance))

print("Remaining features:",
      X_variance.shape[1])


# ================================================================
# 8. CORRELATION FILTER
# ================================================================

print("\n" + "=" * 60)
print("FEATURE SELECTION - CORRELATION FILTER")
print("=" * 60)

corr_matrix = X_variance.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

correlation_threshold = 0.90

high_corr_features = [
    column
    for column in upper_triangle.columns
    if any(
        upper_triangle[column] >
        correlation_threshold
    )
]

X_selected = X_variance.drop(
    columns=high_corr_features
)

print("Before correlation filter:",
      X_variance.shape[1])

print("Highly correlated removed:",
      len(high_corr_features))

print("Final selected features:",
      X_selected.shape[1])


# ================================================================
# 9. DISPLAY FINAL FEATURES
# ================================================================

print("\n" + "=" * 60)
print("FINAL SELECTED FEATURES")
print("=" * 60)

for i, feature in enumerate(
    X_selected.columns,
    1
):
    print(f"{i}. {feature}")


# ================================================================
# 10. SAVE SELECTED FEATURES
# ================================================================

selected_features_df = pd.concat(
    [
        features_df[["customer_id"]],
        X_selected
    ],
    axis=1
)

selected_features_df.to_csv(
    "outputs/feature_selection/selected_features.csv",
    index=False
)


# ================================================================
# 11. CREATE FEATURE SELECTION REPORT
# ================================================================

report = pd.DataFrame({
    "feature": X.columns,

    "passed_variance_filter": [
        f in variance_features
        for f in X.columns
    ],

    "removed_by_correlation": [
        f in high_corr_features
        for f in X.columns
    ],

    "final_selected": [
        f in X_selected.columns
        for f in X.columns
    ]
})

report.to_csv(
    "outputs/feature_selection/feature_selection_report.csv",
    index=False
)


# ================================================================
# 12. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print(
    "Original features       :",
    X.shape[1]
)

print(
    "After variance filtering:",
    X_variance.shape[1]
)

print(
    "After correlation filter:",
    X_selected.shape[1]
)

print("\nFiles created:")

print(
    "1. outputs/customer_features.csv"
)

print(
    "2. outputs/feature_selection/selected_features.csv"
)

print(
    "3. outputs/feature_selection/feature_selection_report.csv"
)

print("\nFeature selection completed successfully!")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch


# ================================================================
# 1. LOAD SELECTED FEATURES
# ================================================================

file_path = "outputs/feature_selection/selected_features.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)

if "customer_id" in df.columns:
    customer_ids = df["customer_id"].copy()
    X = df.drop(columns=["customer_id"])
else:
    customer_ids = pd.Series(
        [f"Customer_{i+1}" for i in range(len(df))]
    )
    X = df.copy()

X = X.select_dtypes(include=np.number)
X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(X.median())

print("Number of customers:", len(X))
print("Number of selected features:", X.shape[1])


# ================================================================
# 2. STANDARDIZE FEATURES
# ================================================================

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# ================================================================
# 3. PCA FOR VISUALIZATION
# ================================================================
# PCA is NOT used to decide the clustering model.
# It is only used to visualize the clusters in 2 dimensions.

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print(
    "PCA variance explained:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)


# ================================================================
# 4. OUTPUT FOLDER
# ================================================================

os.makedirs("outputs/clustering", exist_ok=True)


# ================================================================
# 5. COMPARE 3 CLUSTERING MODELS
# ================================================================

results = []
all_labels = {}

# Test the requested business range: 3 to 5 clusters
k_values = [3, 4, 5]


for k in k_values:

    # ------------------------------------------------------------
    # MODEL 1: K-MEANS
    # ------------------------------------------------------------

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    kmeans_labels = kmeans.fit_predict(X_scaled)

    silhouette = silhouette_score(
        X_scaled,
        kmeans_labels
    )

    db = davies_bouldin_score(
        X_scaled,
        kmeans_labels
    )

    ch = calinski_harabasz_score(
        X_scaled,
        kmeans_labels
    )

    results.append({
        "model": "K-Means",
        "n_clusters": k,
        "silhouette_score": silhouette,
        "davies_bouldin_score": db,
        "calinski_harabasz_score": ch
    })

    all_labels[f"KMeans_{k}"] = kmeans_labels


    # ------------------------------------------------------------
    # MODEL 2: HIERARCHICAL / AGGLOMERATIVE
    # ------------------------------------------------------------

    hierarchical = AgglomerativeClustering(
        n_clusters=k,
        linkage="ward"
    )

    hierarchical_labels = hierarchical.fit_predict(
        X_scaled
    )

    silhouette = silhouette_score(
        X_scaled,
        hierarchical_labels
    )

    db = davies_bouldin_score(
        X_scaled,
        hierarchical_labels
    )

    ch = calinski_harabasz_score(
        X_scaled,
        hierarchical_labels
    )

    results.append({
        "model": "Hierarchical",
        "n_clusters": k,
        "silhouette_score": silhouette,
        "davies_bouldin_score": db,
        "calinski_harabasz_score": ch
    })

    all_labels[f"Hierarchical_{k}"] = hierarchical_labels


    # ------------------------------------------------------------
    # MODEL 3: GAUSSIAN MIXTURE MODEL
    # ------------------------------------------------------------

    gmm = GaussianMixture(
        n_components=k,
        covariance_type="full",
        random_state=42,
        n_init=10
    )

    gmm_labels = gmm.fit_predict(X_scaled)

    silhouette = silhouette_score(
        X_scaled,
        gmm_labels
    )

    db = davies_bouldin_score(
        X_scaled,
        gmm_labels
    )

    ch = calinski_harabasz_score(
        X_scaled,
        gmm_labels
    )

    results.append({
        "model": "Gaussian Mixture",
        "n_clusters": k,
        "silhouette_score": silhouette,
        "davies_bouldin_score": db,
        "calinski_harabasz_score": ch
    })

    all_labels[f"GMM_{k}"] = gmm_labels


# ================================================================
# 6. RESULTS TABLE
# ================================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="silhouette_score",
    ascending=False
)

print("\n" + "=" * 70)
print("CLUSTERING MODEL COMPARISON")
print("=" * 70)

print(
    results_df.to_string(index=False)
)

results_df.to_csv(
    "outputs/clustering/clustering_model_comparison.csv",
    index=False
)


# ================================================================
# 7. IDENTIFY THE HIGHEST-SILHOUETTE CONFIGURATION
# ================================================================

best_row = results_df.iloc[0]

best_model = best_row["model"]
best_k = int(best_row["n_clusters"])

print("\nBest silhouette configuration:")
print("Model:", best_model)
print("Clusters:", best_k)
print("Silhouette:", round(
    best_row["silhouette_score"], 4
))
print("Davies-Bouldin:", round(
    best_row["davies_bouldin_score"], 4
))
print("Calinski-Harabasz:", round(
    best_row["calinski_harabasz_score"], 2
))


# ================================================================
# 8. GET FINAL LABELS
# ================================================================

label_key = {
    "K-Means": f"KMeans_{best_k}",
    "Hierarchical": f"Hierarchical_{best_k}",
    "Gaussian Mixture": f"GMM_{best_k}"
}[best_model]

final_labels = all_labels[label_key]

clustered_df = df.copy()
clustered_df["cluster"] = final_labels

clustered_df.to_csv(
    "outputs/clustering/final_cluster_assignments.csv",
    index=False
)


# ================================================================
# 9. CLUSTER SIZE
# ================================================================

cluster_sizes = (
    clustered_df["cluster"]
    .value_counts()
    .sort_index()
    .rename_axis("cluster")
    .reset_index(name="customer_count")
)

cluster_sizes["percentage"] = (
    cluster_sizes["customer_count"]
    / len(clustered_df)
    * 100
)

print("\nCluster sizes:")
print(cluster_sizes.to_string(index=False))

cluster_sizes.to_csv(
    "outputs/clustering/cluster_sizes.csv",
    index=False
)


# ================================================================
# 10. PCA CLUSTER VISUALIZATION
# ================================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=final_labels,
    alpha=0.75
)

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)"
)
plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)"
)

plt.title(
    f"Final Customer Clusters - {best_model} ({best_k} clusters)"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.tight_layout()

plt.savefig(
    "outputs/clustering/final_clusters_pca.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 11. MODEL COMPARISON GRAPHS
# ================================================================

# ---- Silhouette Score ----

plt.figure(figsize=(10, 6))

for model in results_df["model"].unique():

    temp = results_df[
        results_df["model"] == model
    ]

    plt.plot(
        temp["n_clusters"],
        temp["silhouette_score"],
        marker="o",
        label=model
    )

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score Comparison")
plt.xticks(k_values)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    "outputs/clustering/silhouette_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ---- Davies-Bouldin Score ----

plt.figure(figsize=(10, 6))

for model in results_df["model"].unique():

    temp = results_df[
        results_df["model"] == model
    ]

    plt.plot(
        temp["n_clusters"],
        temp["davies_bouldin_score"],
        marker="o",
        label=model
    )

plt.xlabel("Number of Clusters")
plt.ylabel("Davies-Bouldin Score")
plt.title("Davies-Bouldin Score Comparison")
plt.xticks(k_values)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    "outputs/clustering/davies_bouldin_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 12. PROCESS FLOW DIAGRAM
# ================================================================

fig, ax = plt.subplots(figsize=(12, 16))

ax.set_xlim(0, 10)
ax.set_ylim(0, 18)
ax.axis("off")

boxes = [
    (5, 17, "Raw Smart-Meter Data\n370 customers"),
    (5, 15.5, "Data Quality Audit\nMissing values • gaps • zero runs • spikes"),
    (5, 14, "Feature Engineering\n30+ customer-level load features"),
    (5, 12.5, "Feature Selection\nVariance + Correlation Filtering"),
    (5, 11, "Standardization\nScale customer features"),
    (5, 9.5, "PCA\n2D visualization / dimensionality reduction"),
    (5, 7.7, "Three Clustering Models\nK-Means • Hierarchical • GMM"),
    (5, 5.8, "Model Evaluation\nSilhouette • Davies-Bouldin • Calinski-Harabasz"),
    (5, 4.0, "Select Cluster Configuration\n3–5 meaningful load types"),
    (5, 2.5, "Customer Load Profiles\nTypical daily / weekly patterns"),
    (5, 1.0, "Business Analysis\nPeak contribution • Tariff • Demand Response • Anomalies")
]

def draw_box(x, y, text, width=7.2, height=0.85):
    box = FancyBboxPatch(
        (x - width / 2, y - height / 2),
        width,
        height,
        boxstyle="round,pad=0.04",
        linewidth=1.5
    )
    ax.add_patch(box)
    ax.text(
        x,
        y,
        text,
        ha="center",
        va="center",
        fontsize=10
    )

for x, y, text_value in boxes:
    draw_box(x, y, text_value)

# Arrows
for i in range(len(boxes) - 1):
    x1, y1, _ = boxes[i]
    x2, y2, _ = boxes[i + 1]

    arrow = FancyArrowPatch(
        (x1, y1 - 0.45),
        (x2, y2 + 0.45),
        arrowstyle="->",
        mutation_scale=18,
        linewidth=1.5
    )

    ax.add_patch(arrow)

ax.set_title(
    "Smart-Meter Load Profiling - Machine Learning Process Flow",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.tight_layout()

plt.savefig(
    "outputs/clustering/process_flow_diagram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\n" + "=" * 70)
print("COMPLETED")
print("=" * 70)

print("\nGenerated files:")

print("outputs/clustering/clustering_model_comparison.csv")
print("outputs/clustering/final_cluster_assignments.csv")
print("outputs/clustering/cluster_sizes.csv")
print("outputs/clustering/final_clusters_pca.png")
print("outputs/clustering/silhouette_comparison.png")
print("outputs/clustering/davies_bouldin_comparison.png")
print("outputs/clustering/process_flow_diagram.png")
```

```python
# ================================================================
# XGBOOST WORKING EXAMPLE + VISUALIZATION
# Classification Example
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from xgboost import XGBClassifier

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    auc
)

# ================================================================
# 1. LOAD DATASET
# ================================================================

data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)

print("Dataset shape:", X.shape)

print("\nClasses:")
print(data.target_names)


# ================================================================
# 2. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])


# ================================================================
# 3. CREATE XGBOOST MODEL
# ================================================================

model = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42
)


# ================================================================
# 4. TRAIN MODEL
# ================================================================

model.fit(
    X_train,
    y_train
)

print("\nXGBoost model trained successfully!")


# ================================================================
# 5. PREDICTION
# ================================================================

y_pred = model.predict(X_test)

y_probability = model.predict_proba(X_test)[:, 1]


# ================================================================
# 6. ACCURACY
# ================================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print("Accuracy:", round(accuracy, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=data.target_names
    )
)


# ================================================================
# 7. VISUALIZATION 1
# CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=data.target_names
)

disp.plot()

plt.title("XGBoost Confusion Matrix")

plt.tight_layout()

plt.show()


# ================================================================
# 8. VISUALIZATION 2
# FEATURE IMPORTANCE
# ================================================================

importance = model.feature_importances_

importance_df = pd.DataFrame({
    "feature": X.columns,
    "importance": importance
})

importance_df = importance_df.sort_values(
    by="importance",
    ascending=False
)

# Top 15 features
top_features = importance_df.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["feature"][::-1],
    top_features["importance"][::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.title(
    "Top 15 XGBoost Feature Importances"
)

plt.tight_layout()

plt.show()


# ================================================================
# 9. VISUALIZATION 3
# ROC CURVE
# ================================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)

roc_auc = auc(
    fpr,
    tpr
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"XGBoost AUC = {roc_auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("XGBoost ROC Curve")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ================================================================
# 10. VISUALIZATION 4
# ACTUAL VS PREDICTED
# ================================================================

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison = comparison.reset_index(drop=True)

plt.figure(figsize=(12, 5))

plt.plot(
    comparison.index[:50],
    comparison["Actual"][:50],
    marker="o",
    label="Actual"
)

plt.plot(
    comparison.index[:50],
    comparison["Predicted"][:50],
    marker="x",
    linestyle="--",
    label="Predicted"
)

plt.xlabel("Test Sample")

plt.ylabel("Class")

plt.title(
    "XGBoost: Actual vs Predicted (First 50 Test Samples)"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ================================================================
# 11. PRINT TOP FEATURES
# ================================================================

print("\n" + "=" * 60)
print("TOP 10 IMPORTANT FEATURES")
print("=" * 60)

print(
    importance_df.head(10).to_string(index=False)
)
```

```python
# ================================================================
# XGBOOST - 5 HYPERPARAMETER CONFIGURATIONS
# Compare model performance
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from xgboost import XGBClassifier

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ================================================================
# 1. LOAD DATASET
# ================================================================

data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(data.target)

print("Dataset shape:", X.shape)


# ================================================================
# 2. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ================================================================
# 3. FIVE HYPERPARAMETER CONFIGURATIONS
# ================================================================

configs = {

    "Model 1 - Baseline": {
        "n_estimators": 100,
        "max_depth": 3,
        "learning_rate": 0.10,
        "subsample": 1.0,
        "colsample_bytree": 1.0
    },

    "Model 2 - More Trees": {
        "n_estimators": 200,
        "max_depth": 3,
        "learning_rate": 0.10,
        "subsample": 1.0,
        "colsample_bytree": 1.0
    },

    "Model 3 - Deeper Trees": {
        "n_estimators": 200,
        "max_depth": 6,
        "learning_rate": 0.05,
        "subsample": 0.90,
        "colsample_bytree": 0.90
    },

    "Model 4 - Regularized": {
        "n_estimators": 300,
        "max_depth": 4,
        "learning_rate": 0.03,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "min_child_weight": 3,
        "gamma": 0.1
    },

    "Model 5 - Strong Regularization": {
        "n_estimators": 400,
        "max_depth": 3,
        "learning_rate": 0.02,
        "subsample": 0.75,
        "colsample_bytree": 0.75,
        "min_child_weight": 5,
        "gamma": 0.2
    }
}


# ================================================================
# 4. TRAIN AND EVALUATE ALL 5 MODELS
# ================================================================

results = []

trained_models = {}

for model_name, params in configs.items():

    print("\n" + "=" * 60)
    print(model_name)
    print("=" * 60)

    model = XGBClassifier(
        **params,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    # Train
    model.fit(
        X_train,
        y_train
    )

    # Prediction
    y_pred = model.predict(X_test)

    # Probability
    y_prob = model.predict_proba(X_test)[:, 1]

    # Metrics
    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred
    )

    recall = recall_score(
        y_test,
        y_pred
    )

    f1 = f1_score(
        y_test,
        y_pred
    )

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": auc
    })

    trained_models[model_name] = model

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))
    print("ROC-AUC  :", round(auc, 4))


# ================================================================
# 5. CREATE COMPARISON TABLE
# ================================================================

results_df = pd.DataFrame(results)

print("\n" + "=" * 80)
print("XGBOOST HYPERPARAMETER COMPARISON")
print("=" * 80)

print(
    results_df.to_string(index=False)
)


# ================================================================
# 6. SAVE RESULTS
# ================================================================

import os

os.makedirs(
    "outputs/xgboost",
    exist_ok=True
)

results_df.to_csv(
    "outputs/xgboost/hyperparameter_comparison.csv",
    index=False
)


# ================================================================
# 7. VISUALIZATION - ACCURACY
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    results_df["Model"],
    results_df["Accuracy"]
)

plt.ylabel("Accuracy")
plt.xlabel("XGBoost Configuration")
plt.title("XGBoost Accuracy Comparison")

plt.ylim(
    max(0, results_df["Accuracy"].min() - 0.05),
    1.0
)

plt.xticks(
    rotation=25,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "outputs/xgboost/accuracy_comparison.png",
    dpi=300
)

plt.show()


# ================================================================
# 8. VISUALIZATION - ALL METRICS
# ================================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]

x = np.arange(len(results_df))
width = 0.15

plt.figure(figsize=(14, 7))

for i, metric in enumerate(metrics):

    plt.bar(
        x + (i - 2) * width,
        results_df[metric],
        width,
        label=metric
    )

plt.xlabel("XGBoost Model")
plt.ylabel("Score")
plt.title(
    "Comparison of 5 XGBoost Hyperparameter Configurations"
)

plt.xticks(
    x,
    [
        "Model 1",
        "Model 2",
        "Model 3",
        "Model 4",
        "Model 5"
    ]
)

plt.ylim(0, 1.05)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "outputs/xgboost/all_metrics_comparison.png",
    dpi=300
)

plt.show()


# ================================================================
# 9. FIND BEST CONFIGURATION
# ================================================================

best_index = results_df["F1 Score"].idxmax()

best_model_name = results_df.loc[
    best_index,
    "Model"
]

print("\n" + "=" * 70)
print("BEST CONFIGURATION BASED ON F1 SCORE")
print("=" * 70)

print(
    results_df.loc[
        best_index
    ].to_string()
)

print("\nHyperparameters:")

print(
    configs[best_model_name]
)
```
